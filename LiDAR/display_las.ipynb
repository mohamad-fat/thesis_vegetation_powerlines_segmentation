```python
import laspy
import numpy as np
import open3d as o3d

# Load LAS file
las_file = "data/las_out/voo_3_electrical_merged.las"
las = laspy.read(las_file)

# Extract XYZ points
points = np.vstack((las.x, las.y, las.z)).transpose()

# Create Open3D point cloud
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

# Optional: Downsample point cloud for faster visualization
downsampled = pcd.voxel_down_sample(voxel_size=0.1)  # adjust voxel size
o3d.visualization.draw_geometries([downsampled])

# Display point cloud
o3d.visualization.draw_geometries([pcd])
```
